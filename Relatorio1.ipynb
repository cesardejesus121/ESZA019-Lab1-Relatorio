### 3.1 Atividade A — Leitura de imagem em arquivo

O programa original lê a imagem `messi5.jpg` e a exibe em uma janela do OpenCV. O ponto principal da atividade é entender por que a imagem é exibida em escala de cinza.


```python
import cv2 as cv

# Código original: o segundo argumento igual a 0 força a leitura em escala de cinza.
img = cv.imread('messi5.jpg', 0)
cv.imshow('image', img)
k = cv.waitKey(0)

if k == 27:  # tecla ESC
    cv.destroyAllWindows()
elif k == ord('s'):
    cv.imwrite('messigray.png', img)
    cv.destroyAllWindows()
```

**Resposta:** a janela não mostra a imagem colorida porque o comando `cv.imread('messi5.jpg', 0)` usa o argumento `0`, equivalente a `cv.IMREAD_GRAYSCALE`. Esse modo descarta a informação dos canais de cor e carrega a imagem com apenas uma matriz de intensidades. Para abrir a imagem colorida, deve-se usar `1` ou `cv.IMREAD_COLOR`.


```python
import cv2 as cv

# Código corrigido: leitura colorida.
img_color = cv.imread('messi5.jpg', cv.IMREAD_COLOR)
cv.imshow('image_colorida', img_color)
k = cv.waitKey(0) & 0xFF

if k == 27:  # tecla ESC
    cv.destroyAllWindows()
elif k == ord('s'):
    cv.imwrite('messi_colorida.png', img_color)
    cv.destroyAllWindows()
```

### 3.2 Atividade B — Leitura de vídeo em arquivo

O programa lê o vídeo `big_buck_bunny.mp4` e exibe seus quadros sequencialmente. A velocidade aparente do vídeo depende do tempo de espera entre um quadro e o próximo.


```python
import time
import cv2 as cv

cap = cv.VideoCapture('big_buck_bunny.mp4')

if not cap.isOpened():
    raise RuntimeError('Não foi possível abrir o vídeo big_buck_bunny.mp4')

fps_video = cap.get(cv.CAP_PROP_FPS)
intervalo_normal = 1 / fps_video if fps_video > 0 else 1 / 25.0

# Para exibição normal, use fator_velocidade = 1.0.
# Para exibição mais rápida, use fator_velocidade > 1. Exemplo: 4.0.
# Para exibição mais lenta, use fator_velocidade < 1. Exemplo: 0.5.
fator_velocidade = 1.0
intervalo = intervalo_normal / fator_velocidade

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    cv.imshow('frame', frame)
    time.sleep(intervalo)

    if cv.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv.destroyAllWindows()
```

**Explicação:** um vídeo é formado por uma sequência de imagens chamadas quadros, ou *frames*. Ao aumentar o tempo de espera entre os quadros, a exibição fica mais lenta. Ao diminuir esse tempo, os quadros são mostrados mais rapidamente e o vídeo aparenta estar acelerado.

No código, esse controle é feito pela variável `intervalo`. Se o vídeo tem aproximadamente 25 quadros por segundo, o intervalo normal entre quadros é próximo de `1/25` segundo. Para acelerar, pode-se usar um intervalo menor, como `1/100` segundo. Para desacelerar, pode-se usar um intervalo maior, como `1/10` ou `1/5` segundo.


### 3.3 Atividade C — Captura de imagem da webcam

Nesta atividade, o programa foi modificado para salvar uma imagem da webcam no arquivo `foto1.png` quando a tecla `x` fosse pressionada.


```python
import cv2 as cv

cap = cv.VideoCapture(0)

if not cap.isOpened():
    raise RuntimeError('Não foi possível abrir a câmera')

print("Pressione 'x' para salvar foto1.png ou 'q' para sair.")

while True:
    ret, frame = cap.read()
    if not ret:
        print('Não foi possível receber o frame. Encerrando...')
        break

    cv.imshow('frame', frame)
    key = cv.waitKey(1) & 0xFF

    if key == ord('x'):
        cv.imwrite('foto1.png', frame)
        print("Imagem salva como 'foto1.png'.")
    elif key == ord('q'):
        break

cap.release()
cv.destroyAllWindows()
```

**Solução:** a variável `key` armazena a tecla pressionada. Quando `key == ord('x')`, o frame atual da câmera é salvo por `cv.imwrite('foto1.png', frame)`. Assim, a imagem registrada corresponde ao quadro que estava sendo visualizado no instante em que a tecla `x` foi pressionada.


### 3.4 Atividade D — Gravação de vídeo da webcam

Nesta atividade, o programa foi ajustado para gravar a imagem normal, sem inversão vertical, e com uma taxa de quadros adequada para reprodução.


```python
import cv2 as cv

cap = cv.VideoCapture(0)

if not cap.isOpened():
    raise RuntimeError('Não foi possível abrir a câmera')

width = int(cap.get(cv.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv.CAP_PROP_FRAME_HEIGHT))

# FPS adequado para uma gravação simples de webcam.
fps = 10.0

fourcc = cv.VideoWriter_fourcc(*'XVID')
out = cv.VideoWriter('saida.avi', fourcc, fps, (width, height))

print("Pressione 'q' para encerrar a gravação.")

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        print('Não foi possível receber o frame. Encerrando...')
        break

    # Qualquer processamento da imagem deve ser feito aqui,
    # depois da leitura do frame e antes da exibição/gravação.

    out.write(frame)
    cv.imshow('frame', frame)

    if cv.waitKey(int(1000 / fps)) & 0xFF == ord('q'):
        break

cap.release()
out.release()
cv.destroyAllWindows()
```

**Solução:** para que a imagem gravada fique normal, foi removida a inversão vertical do frame. Também foi escolhida uma taxa de quadros de `10 fps`, suficiente para uma gravação simples de webcam e compatível com os vídeos gerados pelo grupo.

**Ponto adequado para processamento:** caso seja necessário aplicar alguma operação de processamento, como conversão para escala de cinza, filtragem, recorte, espelhamento ou detecção de bordas, essa operação deve ser realizada logo após a leitura do frame (`ret, frame = cap.read()`) e antes de `cv.imshow`, `cv.imwrite` ou `out.write`. Isso garante que a imagem processada seja a mesma exibida ou salva.
